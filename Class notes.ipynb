# Recursion

```python
# base case
# resursing case
# factorial(5) 5*4*3*2*1
```

```python
def factorial(n):
    #base case
    if n==1:
           return 1
    # recursive case
    else:
        return n* factorial(n-1)
```

```python
factorial(5)
```

# Lambda Fuction
######anonymous
######small one liner function
######lambda
######alternate of def

```python
add=lambda x,y:x+y
print(add(5,6))
```

```python
numbers=[1,2,3,4]
doubled= map(lambda x:x*2,numbers)
print(list(doubled))
```

```python
numbers=[1,2,3,4,5,6,7]
# display all the even numbers
evens = filter(lambda x:x%2==0,numbers)
print(list(evens))
```
